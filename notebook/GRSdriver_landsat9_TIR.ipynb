# Loading Landsat OLI and TIRS bands on common grid (i.e., same spatial resolution)

```python
import glob
import os
import numpy as np

import pandas as pd
import geopandas as gpd
import xarray as xr
import rioxarray

import matplotlib.pyplot as plt
import matplotlib as mpl
import cartopy.crs as ccrs
from matplotlib.colors import ListedColormap

import GRSdriver
from GRSdriver import visual

opj = os.path.join

print(f'-GRSdriver: {GRSdriver.__version__}')
```

Set the path of the images you want to load, here is a example for the former (v3) and current (v4) image formatting 

```python
image ='/data/satellite/landsat/LC09_L1TP_196030_20221120_20230321_02_T1'
```

```python
# set resolution to 100 m, native resolution of TIR bands (recommended)
l1c = GRSdriver.LandsatDriver(image,band_idx=[1,2,4,9,10],resolution=100)
```

```python
epsg = l1c.prod.rio.crs.to_epsg()
str_epsg = str(epsg)
zone = str_epsg[-2:]
is_south = str_epsg[2] == 7
proj = ccrs.UTM(zone, is_south)

coarsening=1
cmap = plt.cm.Spectral_r
fig = plt.figure(figsize=(20, 15))
#ax = plt.subplot(1, 1, 1, projection=proj)
#ax.set_extent(extent_val, proj)
l1c.prod.bands.isel(wl=[0,1,2]).plot.imshow(col='wl',col_wrap=3, cmap=cmap,robust=True,subplot_kws=dict(projection=proj))
```

```python
fig = plt.figure(figsize=(20, 15))
l1c.prod.bands.isel(wl=-1).plot.imshow(cmap=cmap,robust=True,subplot_kws=dict(projection=proj))
```
