Set the path of the images you want to load, here is a example for the former (v3) and current (v4) image formatting 

```python
image ='/data/satellite/Sentinel-2/L1C/20PNC/2023/03/17/S2B_MSIL1C_20230317T144729_N0509_R139_T20PNC_20230317T175159.SAFE'
image = '/data/tmp/S2B_MSIL1C_20230317T144729_N0509_R139_T20PNC_20230317T175159.nc'
cams_file = '/data/cams/world/cams_forecast_2023-03.nc'
```

```python
raster = xr.open_dataset(image)
```

Open CAMS data with lazy loading

```python
cams =xr.open_dataset(cams_file)
cams
```

Get data for the date of the image

```python
date = raster.time.dt.strftime('%Y-%m-%d').values
date
```

Now you can subset to the satellite image extent

```python
cams.rio.bounds()
```

```python
xmin, ymin, xmax, ymax = raster.rio.bounds()
lonmin, latmin, lonmax, latmax = raster.rio.transform_bounds(4326)

cams = cams.sel(latitude=slice(latmax + 1, latmin - 1))
# check if image is on Greenwich meridian and adapt longitude convention
if cams.longitude.min()>=0:
    if lonmin <= 0 and lonmax >= 0:

        cams = cams.assign_coords({"longitude": (((cams.longitude + 180) % 360) - 180)}).sortby('longitude')
    else:
        # set longitude between 0 and 360 deg
        lonmin, lonmax, = lonmin % 360, lonmax % 360

# slicing
cams = cams.sel(longitude=slice(lonmin - 1, lonmax + 1))
#.compute()
cams
```

if your subset is small, it might be useful to interpolate within the native raster

```python
regrid_number=500
cams=cams.interp(latitude=np.linspace(cams.latitude.max(),cams.latitude.min(),regrid_number),
            longitude=np.linspace(cams.longitude.min(),cams.longitude.max(),regrid_number),method='cubic')
cams
```

```python
(cams.aod550).plot.imshow(col='time',col_wrap=4,aspect=1.1,cmap=plt.cm.Spectral_r,robust=True,subplot_kws=dict(projection=proj),cbar_kwargs={'shrink':0.5}) #,vmin=-5,vmax=20)
```

```python
fig, axs = plt.subplots(1,2, figsize=(15, 8),subplot_kw={'projection':proj})

(cams.aod550.mean('time')).plot.imshow(ax=axs[0],cmap=plt.cm.Spectral_r,robust=True,cbar_kwargs={'shrink':0.5}) #,vmin=-5,vmax=20)
(cams.aod550.std('time')).plot.imshow(ax=axs[1],cmap=plt.cm.Spectral_r,robust=True,cbar_kwargs={'shrink':0.5}) #,vmin=-5,vmax=20)
```

```python
params=['aod550','amaod550', 'bcaod550', 'duaod550', 'niaod550', 'omaod550', 'ssaod550', 'suaod550']
Nrows = (len(params) + 4) // 4
fig, axs = plt.subplots(Nrows, 4, figsize=(4 * 4.2, Nrows * 3.5))
axs = axs.ravel()
[axi.set_axis_off() for axi in axs]
aod550 = cams['aod550']
aod550
for i, param in enumerate(params):
    if i == 0:
        fig = cams[param].plot.imshow(robust=True, ax=axs[i],cmap=plt.cm.Spectral_r)
        fig.axes.set_title(param)
    else:
        fig = (cams[param]/aod550).plot.imshow(robust=True, ax=axs[i],cmap=plt.cm.Spectral_r)
        fig.axes.set_title('prop. of '+param)
    #fig.colorbar.set_label(self.raster[param].units)
    fig.axes.set(xticks=[], yticks=[])
    fig.axes.set_ylabel('')
    fig.axes.set_xlabel('')
plt.tight_layout()
```
